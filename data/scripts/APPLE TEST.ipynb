На основе сырых описаний необходимо получить следующие признаки (характеристики):

- **apple_variety**: сорт яблок (например, "АНТОНОВКА", "ГРЕННИ СМИТ" и т.п.)
- **apple_color**: цвет яблок, один из следующих: "ЖЕЛТЫЙ", "ЗЕЛЕНЫЙ", "КРАСНЫЙ"
- **apple_package**: тип упаковки: "ВЕСОВОЕ" (продается на вес), "ФАСОВАННОЕ" (упаковано в пакеты/коробки)
- **apple_weight**: вес яблок в килограммах (числовое значение, выделить из описания; может быть в граммах или килограммах, требуется нормализация)
- **apple_keyword**: признак наличия вариации ключевого слова "ЯБЛОКИ" в сыром описании в описании ("наличие" или "отсутствие")<br><br>
